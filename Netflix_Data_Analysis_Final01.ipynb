# Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to identify hidden patterns, trends, and relationships within the Netflix dataset. Various visualizations are used to answer business-related questions.

## Business Question 1

### What is the distribution of Movies and TV Shows available on Netflix?

Understanding the distribution helps identify Netflix's primary content strategy and whether the platform focuses more on Movies or TV Shows.

```python
df["type"].value_counts()
```

```python
sns.countplot(x="type", data=df)
plt.title("Distribution of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Count")
plt.show()
```

### Interpretation

The graph compares the number of Movies and TV Shows available on Netflix.

Movies are significantly higher in number than TV Shows, indicating that Netflix has a stronger focus on movie content.

### Business Insight

Netflix invests more heavily in Movies than TV Shows. This suggests that Movies form the core of Netflix's content library and may contribute more to viewer engagement.

## Business Question 2

### Which countries contribute the highest amount of content on Netflix?

Analyzing content by country helps identify Netflix's major content-producing regions and understand its global distribution strategy.

```python
top_country = df["country"].value_counts().head(10)

plt.figure(figsize=(10,6))
top_country.plot(kind="barh")
plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

### Interpretation

The graph displays the top ten countries contributing the highest number of Movies and TV Shows on Netflix.

The United States contributes the largest amount of content, followed by other countries such as India and the United Kingdom.

## Business Question 3

### Which content ratings are most common on Netflix?

Analyzing content ratings helps understand the target audience of Netflix and identify which age groups receive the most content.

```python
top_rating = df["rating"].value_counts().head(10)

plt.figure(figsize=(10,6))
top_rating.plot(kind="bar")
plt.title("Top 10 Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The graph shows the ten most common content ratings available on Netflix.

Ratings such as TV-MA, TV-14, and TV-PG appear most frequently, indicating that Netflix offers a large amount of content for mature and teenage audiences.

### Business Insight

Netflix mainly focuses on content suitable for adults and teenagers, which reflects the preferences of a large portion of its subscriber base.

## Business Question 4

### How has Netflix content changed over the years?

This analysis helps identify trends in content production and understand Netflix's growth over time.

```python
year_data = df.groupby("release_year")["type"].count()

plt.figure(figsize=(12,6))
year_data.plot()
plt.title("Netflix Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

### Interpretation

The line graph shows the number of Movies and TV Shows released each year.

The number of titles increases significantly after 2000, indicating rapid growth in Netflix's content library.

### Business Insight

Netflix has consistently expanded its content catalogue over the years, especially in recent decades, to meet increasing global demand.

## Business Question 5

### Which genres are most popular on Netflix?

Understanding genre distribution helps identify the types of content that Netflix offers most frequently.

```python
top_genres = df["listed_in"].value_counts().head(10)

plt.figure(figsize=(12,6))
top_genres.plot(kind="barh")
plt.title("Top 10 Genres on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()
```

### Interpretation

The graph displays the ten most common genres available on Netflix.

Genres such as Drama, International Movies, and Comedies dominate the Netflix library.

### Business Insight

Netflix provides a diverse range of entertainment, but drama and comedy remain the most frequently offered genres due to their high audience demand.

## Business Question 6

### What is the distribution of movie durations?

Analyzing movie duration helps understand the typical length of movies available on Netflix.

```python
movies = df[df["type"] == "Movie"].copy()

movies["duration"] = movies["duration"].str.replace(" min", "", regex=False)
movies["duration"] = pd.to_numeric(movies["duration"])

plt.figure(figsize=(10,6))
sns.histplot(movies["duration"], bins=20)
plt.title("Distribution of Movie Duration")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Frequency")
plt.show()
```

### Interpretation

The histogram shows the distribution of movie durations.

Most Netflix movies have a duration between approximately 80 and 120 minutes.

### Business Insight

Netflix mainly offers standard-length feature films, suggesting that viewers generally prefer movies within the 80–120 minute range.

## Business Question 7

### Which directors have contributed the highest number of titles on Netflix?

This analysis identifies the directors with the largest number of Movies or TV Shows available on Netflix.

```python
top_directors = df["director"].value_counts().head(10)

plt.figure(figsize=(12,6))
top_directors.plot(kind="barh")
plt.title("Top 10 Directors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Director")
plt.show()
```

### Interpretation

The graph displays the top ten directors with the highest number of titles on Netflix.

These directors have contributed significantly to Netflix's content library.

### Business Insight

Directors with multiple titles indicate strong partnerships or consistent content acquisition by Netflix, helping maintain a diverse catalogue for viewers.

## Business Question 8

### Which release years have the highest number of titles on Netflix?

This analysis identifies the release years with the highest number of Movies and TV Shows available on Netflix.


```python
top_years = df["release_year"].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_years.index,
    y=top_years.values
)

plt.title("Top 10 Release Years by Number of Titles")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

### Interpretation

The graph displays the ten release years with the highest number of Movies and TV Shows available on Netflix.

It shows that Netflix contains a larger collection of recently released content compared to older releases.

### Business Insight

Netflix continuously updates its content library by acquiring or producing newer titles. Recent release years dominate the platform, reflecting changing audience preferences for modern content.

## Business Question 9

### How has Netflix expanded its content library over time?

This analysis examines the number of titles added to Netflix each year.

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

content_added = df["date_added"].dt.year.value_counts().sort_index()

plt.figure(figsize=(12,6))
content_added.plot(kind="line", marker="o")
plt.title("Content Added to Netflix by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles Added")
plt.show()
```

### Interpretation

The line chart shows how Netflix has added content over the years.

A significant increase in content additions can be observed in recent years.

### Business Insight

Netflix has rapidly expanded its content library, demonstrating continuous investment in new Movies and TV Shows.

## Business Question 10

### What percentage of Netflix content consists of Movies and TV Shows?

A pie chart provides a clear representation of the proportion of Movies and TV Shows available on Netflix.

```python
type_count = df["type"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    type_count,
    labels=type_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Movies vs TV Shows")
plt.show()
```

### Interpretation

The pie chart illustrates the percentage distribution of Movies and TV Shows.

Movies occupy a larger share of Netflix's content library compared to TV Shows.

### Business Insight

Netflix's strategy is more focused on Movies, while maintaining a smaller but important collection of TV Shows.
